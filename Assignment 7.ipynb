```python
#PART C
#bar chart
df[7].value_counts().plot(kind='bar')
plt.title("active vs inactive")
plt.show()
```

Explanation:
shows how many active vs inactive members. good because it’s simple category data


```python
#pie chart
df[2].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("gender split")
plt.ylabel("")
plt.show()
```

Explanation:
shows percent of each gender. pie chart works because there are only 2 groups



```python
#convert column 3 to numeric, replacing non-numeric values with NaN 
df[3] = pd.to_numeric(df[3], errors='coerce')
```

```python
#remove NaN values before plotting 
numeric_data = df[3].dropna()
```

```python
# Create histogram
numeric_data.plot(kind='hist', bins=10)
plt.title("age distribution")
plt.show()
```

Explanation:
shows how ages are spread out. most people look like they are in the middle age range
